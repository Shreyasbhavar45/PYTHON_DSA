```python
#To print square pattern with provided fixed digit
def square(n):
    for i in range(n):
        print((str(n)+" ")*n)
        
square(6)
```

```python
###above same just increasing number with column
def square(n):
    for i in range(n):
        for j in range(n):
            #why we use (J+1)-> because we have to increase the number in column 
            
            print(str(j+1),end = ' ')
        print()

square(int(input("Enter The Number")))
```
