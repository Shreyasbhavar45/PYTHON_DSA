## __1.NODE BASICS__


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

##create nodes
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

##connect node
node1.next = node2
node2.next = node3

##head
head = node1

##Traversal
temp = head
while temp:
    print(temp.data)
    temp = temp.next
```

## __2.COUNT TOTAl NODES__

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

##create a nodes
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

##connect the nodes
node1.next = node2
node2.next = node3

#head
head = node1

#traversal and count
count = 0
temp = head

while temp:
    print(temp.data)
    count = count + 1
    temp = temp.next

print(count)
```

## __3.Search Element__

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

##create a nodes
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

##connect the nodes
node1.next = node2
node2.next = node3

#head
head = node1

##Traverse and search
Target = 20
temp = head

while temp:
    if temp.data == Target:
        print("Found")
        break
        
    temp = temp.next
```

## __4.Sum of Nodes__

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

##create a nodes
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

##connect the nodes
node1.next = node2
node2.next = node3

#head
head = node1

##Traverse and Sum
Total = 0
temp = head

while temp:
    Total = Total + temp.data
    temp = temp.next

print("The sum of the total nodes:-",Total)
```

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.nextPtr = None

class LinkedList:
    
    def __init__(self):
        self.head = None

    ## insert at the new node at the beginning of the LinkedList.
    def insertAtbeginning(self,new_data):
        ## create a new node for the new data
        new_node = Node(new_data)

        ## insertion at the beginning of the Linked List.
        new_node.nextPtr = self.head
        self.head = new_node

    ## print the Linked List
    def printLinkedList(self):
        temp = self.head
        while temp:
            print(str(temp.data) + " ", end = " ")
            temp = temp.nextPtr
        
## Driver Code
llist = LinkedList()
llist.insertAtbeginning(10)
llist.insertAtbeginning(20)
llist.insertAtbeginning(30)
llist.insertAtbeginning(40)
llist.insertAtbeginning(50)
llist.printLinkedList()
```
